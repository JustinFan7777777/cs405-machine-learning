### 1. Array creation and basic operations

```python
# (1) Create a 10x10 random array with values between 0 and 100 (int)

# [0, 101) because the upper bound is exclusive
arr = np.random.randint(0, 101, size = (10, 10))

print(arr)
```

```python
# (2) Calculate the minimum, maximum, and mean of the array

min_value = np.min(arr) # Minimum value
max_value = np.max(arr) # Maximum value

# the mean value is calculated by summing up
# all the elements in the array
# and dividing by the total number of elements.
mean_value = np.mean(arr) # Mean value

print("Min:", min_value, "Max:", max_value, "Mean:", mean_value)
```

### 2. Array reshaping and merging

```python
# (1) Reshape the array into a 5x20 array

# the total number of the elements in the reshaped array
# must be the same as the original array
reshaped_array = arr.reshape(5, 20)

print("Reshaped array:", reshaped_array)
```

```python
# (2) Create a 1x20 array using np.ones, and concatenate it to the reshaped array along the rows

# the new array has 1 row & 20 columns, with all the values being 1
new_array = np.ones((1, 20))

# reshaped_array: (5, 20)
# new_array: (1, 20)
# concatenated_array: (6, 20)
# axis=0 means "moving along the rows"
# requirement: the number of columns in the two arrays must be consistent
concatenated_array = np.concatenate([reshaped_array, new_array], axis=0)

print("Concatenated array:", concatenated_array)
```

### 3. Logical operations and filtering

```python
# (1) Use boolean indexing to find all elements greater than 50, and set them to 50

# np.where(condition, value_if_true, value_if_false)
# condition is true -> value = value_if_true
# condition is false -> value = value_if_false
# arr means every element in the array
arr = np.where(arr > 50, 50, arr)

print("Array after setting elements > 50 to 50:", arr)
```

### 4. Linear algebra operations

```python
# (1) Create a 3x3 identity matrix

# identity matrix: a square matrix with 1s on the main diagonal and 0s elsewhere
identity_matrix = np.eye(3)
print("Identity matrix:", identity_matrix)
```

```python
# (2) Create a 3x3 random matrix and compute its inverse

# np is the defined alias for the numpy library
# random is the module in numpy
# rand is the function in the "random" module
random_matrix = np.random.rand(3, 3)

# linalg refers to "linear algebra"
inv_matrix = np.linalg.inv(random_matrix) # Inverse of the matrix

print("Random matrix:", random_matrix)
print("Inverse matrix:", inv_matrix)
```

```python
# (3) Matrix multiplication

# np.dot(random_matrix, inv_matrix) is also valid in this case
# but the "@" operator is more concise and easier to read for matrix multiplication
matrix_product = random_matrix @ inv_matrix

print("Matrix product (should be close to identity matrix):", matrix_product)
```

#### 1.2.3 Handling the Null/Missing Values

```python
# use isnull().sum() to check for missing values 
df.isnull().sum() 
# isnull(): a pandas method that detects missing values in the DataFrame
# sum(): a pandas method that sums up the number of missing values in each column of the DataFrame
# for example, we can see that there are 22 missing values in the 'LoanAmount' column
# which indicates that some clients did not provide their loan amount information.
```

Missing values exist for **Gender, Married, Dependents, Self_Employed, LoanAmount, Loan_Amount_Term and Credit_History**.

The null values in the dataset are imputed using mean/median or mode based on the type of data that is missing:

- **Numerical Data:** If a numerical value is missing, then replace that NaN value with mean or median. 
- **Categorical Data:** When categorical data is missing, replace that with the value which is most occurring i.e. by mode. 



```python
# fill the missing values for categoriacal terms - mode
df.fillna({'Gender': df['Gender'].value_counts().idxmax()}, inplace=True)
df.fillna({'Married': df['Married'].value_counts().idxmax()}, inplace=True)
df.fillna({'Dependents': df['Dependents'].value_counts().idxmax()}, inplace=True)
df.fillna({'Self_Employed': df['Self_Employed'].value_counts().idxmax()}, inplace=True)
# value_counts().idxmax() is used to find the most frequent value (the mode) in each of the specified categorical columns
# and fill the missing values in those columns with the mode using the fillna() method
# for example, in the 'Gender' column, the most frequent value is 'Male', so we fill the missing values in the 'Gender' column with 'Male'

# fill the missing values for numerical term s -mean or median
df.fillna({'LoanAmount': df["LoanAmount"].mean(skipna=True)}, inplace=True)
# mean(skipna=True) is used to calculate the mean of the 'LoanAmount' column while ignoring the missing values (NaN)
# this is done to fill the missing values in the 'LoanAmount' column with the mean value of the existing non-missing values in that column
# for example, if the mean of the non-missing values in the 'LoanAmount' column is 150, then we fill all the missing values in that column with 150.

df.fillna({'Loan_Amount_Term': df['Loan_Amount_Term'].value_counts().idxmax()}, inplace=True)
df.fillna({'Credit_History': df['Credit_History'].value_counts().idxmax()}, inplace=True)

#Checking missing value again:
df.isnull().sum() 
# check again after filling all the missing values
# now that there are no missing values anymore, we can proceed with further analysis
```

From the above output, we see that there’re no null values and now we can perform the data visualization.

NOTICE: If a column has, let’s say, 50% of its values missing, then do we replace all of those missing values with the respective median or mode value? Actually, we don’t. We delete that particular column in that case. 

Sk-learn library has an in-built function called Iterative Imputer to impute the missing values. Its sklearn domcumentation:[ https://scikit-learn.org/stable/modules/generated/sklearn.impute.IterativeImputer.html](https://scikit-learn.org/stable/modules/impute.html)

#####  1.2.5.2 Using Capping Approach

###### <font size=4> 1) Z-Score approach to treat Outliers:</font>

All the values above 3 standard deviation and below -3 standard deviation are outliers and can be removed.

**Using SciPy Library to calculate the Z-Score:**

```python
# 'SciPy' is used to perform scientific computations
import scipy.stats as stats
# stats: a module in the SciPy library that provides a wide range of statistical functions and tests for analyzing data

# Creating new variable with Z-score of each record:
df2 = df.copy()
# now we get another copy 

df2['ZR'] = stats.zscore(df2['LoanAmount'])
# we create a new column called 'ZR' in the df2 DataFrame
# by applying the z-score transformation to the 'LoanAmount' column using the stats.zscore() function
# the z-score transformation standardizes the 'LoanAmount' variable by subtracting the mean and dividing by the standard deviation
# this transformation is often used to identify outliers in the data, as z-scores greater than 3 or less than -3 are typically considered outliers

df2.head()
```

```python
# Combined Lower limit and Upper limit:
df2[(df2['ZR']<-3) | (df2['ZR']>3)]
# we filter the df2 DataFrame to identify records where the z-score ('ZR') is less than -3 or greater than 3
# this helps us identify potential outliers in the 'LoanAmount' variable based on the z-score transformation.

# we see that there are 13 records in the dataset that have z-scores less than -3 or greater than 3
# which indicates that there are 13 potential outliers in the 'LoanAmount' variable based on the z-score transformation.
```

```python
df2[(df2['ZR']<-3) | (df2['ZR']>3)].shape[0]
# we use the shape attribute to get the number of rows in the filtered DataFrame
# which is exactly the number of potential outliers
```

```python
### Cleaned Data: without outliers so z>-3 and z< +3

df2 = df2[(df2['ZR']>-3) & (df2['ZR']<3)].reset_index()
# we filter the df2 DataFrame to include only records where the z-score ('ZR') is greater than -3 and less than 3
# this effectively removes the potential outliers from the dataset
# reset_index(): a pandas method that resets the index of the DataFrame after filtering
# which is useful to maintain a clean and sequential index for the remaining records in df2.

df2.head()
# we can see that the potential outliers have been removed from the dataset
# and we now have a cleaned version of the data in df2.
```

###### <font size=4> 2) IQR Method to treat Outliers:</font>

All the values below Q1 - 1.5*IQR and values above Q3 + 1.5*IQR are outliers and can be removed.

```python
# finding the Quantiles:

Q1 = df3.LoanAmount.quantile(0.25)
Q2 = df3.LoanAmount.quantile(0.50)
Q3 = df3.LoanAmount.quantile(0.75)
# quantile(): a pandas method that calculates the specified quantiles of the 'LoanAmount' column
# Q1: the 25th percentile, Q2: the 50th percentile (also known as the median), and Q3: the 75th percentile of the 'LoanAmount' variable

# IQR : Inter-Quartile Range

IQR = Q3 - Q1
# IQR is calculated as the difference between the 75th percentile (Q3) and the 25th percentile (Q1)
# IQR is a measure of statistical dispersion and is used to identify potential outliers in the data

# Lower Limit:
LC = Q1 - (1.5*IQR)
# the lower limit (LC) is calculated as the 25th percentile (Q1) minus 1.5 times the inter-quartile range (IQR)
# any data point that is less than the lower limit (LC) is considered a potential outlier

# Upper Limit:
UC = Q3 + (1.5*IQR)
# the upper limit (UC) is calculated as the 75th percentile (Q3) plus 1.5 times the inter-quartile range (IQR)
# any data point that is greater than the upper limit (UC) is considered a potential outlier

display(LC)
display(UC)
# we display the calculated lower limit (LC) and upper limit (UC) for the 'LoanAmount' variable
```

```python
## Plot

sns.histplot(df3.LoanAmount)
plt.axvline(UC, color='r')
plt.axvline(LC, color ='r')
plt.axvline(Q1, color='g')
plt.axvline(Q3, color='g')
# axvline(): a matplotlib function that adds a vertical line to the plot at the specified x-coordinate
# we add vertical lines to the histogram to indicate the upper limit (UC), lower limit (LC)
# and the first and third quartiles (Q1 and Q3) of the 'Loan
# UC and LC are shown in red, while Q1 and Q3 are shown in green on the histogram

plt.show()
```

```python
# Find count of Outliers wrt IQR

df3[(df3.LoanAmount<LC) | (df3.LoanAmount>UC)].reset_index(drop=True)
# we filter the df3 DataFrame by including only records where the 'LoanAmount' is less than the lower limit (LC) or greater than the upper limit (UC)
# reset_index(drop=True): a pandas method that resets the index of the filtered DataFrame and drops the old index
```

```python
df3[(df3.LoanAmount<LC) | (df3.LoanAmount>UC)].shape[0]
# number of rows in the filtered DataFrame gives us the count of potential outliers in the 'LoanAmount' variable based on the IQR method
```

### 1) Accuracy
  
<font color=red> Accuracy</font> measures how often the classifier makes the correct prediction. It’s the ratio of the number of correct predictions to the total number of predictions.
$$ACC = \frac{TP+TN}{TP+FP+TN+FN}$$

<br/>
In Python you can calculate it in the following way:

```python
from sklearn.metrics import confusion_matrix, accuracy_score
y_pred = model.predict_proba(X_test)#Return probability estimates for the test vector X.
# we feed in the test datato the model and get the results in the form of probabilities for each class
# (e.g., probability of loan approval and probability of loan rejection)

threshold = 0.8
# we set a threshold value of 0.8 to classify the predictions into binary classes (e.g., approved or rejected)

y_pred_class = y_pred[:, 1]   > threshold
# [:, 1]: we select the probabilities for the positive class (e.g., loan approval)
# > threshold: we compare the probabilities with the threshold value to classify the predictions into binary classes
# if the probability of loan approval is greater than 0.8, we classify it as True (approved)
# otherwise we classify it as False (rejected)

tn, fp, fn, tp = confusion_matrix(y_test, y_pred_class).ravel()
# confusion_matrix(): a function from the sklearn.metrics module that computes the confusion matrixx
# to evaluate the performance of a classification model
# ravel(): a method that flattens the confusion matrix into a 1D array
# which allows us to unpack the true negatives (tn), false positives (fp), false negatives (fn), and true positives (tp) into separate variables

accuracy = (tp + tn) / (tp + fp + fn + tn)
# we calculate the accuracy based on the ACC formula
# (True Positives + True Negatives) / (Total Predictions)

# Or simply
accuracy_score(y_test, y_pred_class)
# we can also use the accuracy_score() function from the sklearn.metrics module to calculate the accuracy directly
```


### 2) Precision: <br/>**Precision** tells us what proportion of messages we classified as positive. It is a ratio of true positives to all positive predictions. In other words, 
$$ Precision = TP/(TP + FP) $$

### 3)  Recall: <br/>**Recall(sensitivity)** tells us what proportion of messages that actually were positive were classified by us as positive.
$$Recall = TP/(TP + FN)$$

### 4)  F1 score: <br/>
We can use **F-beta** score as a metric that considers both precision and  recall:
$$F_\beta=(1+\beta^2)\bullet\frac{precision\bullet recall}{(\beta^2 \bullet precision)+recall}$$
When choosing beta in your F-beta score **the more you care about recall** over precision **the higher beta** you should choose. For example, with **[F1 score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html#sklearn.metrics.f1_score) we care equally about recall and precision** with F2 score, recall is twice as important to us.

### 5)  TPR & FPR & ROC & AUC: 


$$ TPR=\frac{positives\ correctly\ classified}{total\ positives}=\frac{TP}{TP+FN}=\frac{TP}{P} $$
$$ FPR=\frac{negatives\ incorrectly\ classified}{total\ negatives}=\frac{FP}{TN+FP}=\frac{FP}{N} $$

ROC（Receiver Operating Characteristic is used to measure the output quality of the evaluation classifier.
ROC curves are two-dimensional graphs in which true positive rate (TPR) is ploted on the Y axis and false positive rate (FPR) is plotted on the X axis. An ROC graph depicts relative tradeoffs between true positive rate (TPR) and false positive rate (FPR). Basically, for every threshold, we calculate TPR and FPR and plot it on one chart.
